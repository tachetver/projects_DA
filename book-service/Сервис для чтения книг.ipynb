## Запросы к базе данных

### 1. Кол-во книг, выпущенных после 1 января 2000 года

```python
query = '''
            SELECT COUNT (DISTINCT book_id) as cnt
            FROM books
            WHERE publication_date > '2000-01-01'
'''
pd.io.sql.read_sql(query, con = engine)
```

**Вывод**: после 1 января 2000 года вышло всего 819 книг

### 2. Количество обзоров и средняя оценка для каждой книги

```python
query = '''
            SELECT books.book_id,
                    books.title,
                    count(distinct reviews.review_id) as cnt_rev,
                    round(avg(ratings.rating), 2) as avg_rat
            FROM books
            left join reviews on reviews.book_id = books.book_id
            left join ratings on ratings.book_id = books.book_id
            GROUP BY books.book_id
            ORDER BY cnt_rev desc, avg_rat desc
'''
pd.io.sql.read_sql(query, con = engine)
```

**Вывод**: больше всего обзоров сделано на книгу Twilight, у 6 книг нет обзоров, самый высокий средний рейтинг (5.00) у 44 книг

### 3. Издательство, которое выпустило наибольшее число книг толще 50 страниц

```python
query = '''
            SELECT publishers.publisher,
            count(books.book_id) as cnt_books
            FROM books
            left join publishers on publishers.publisher_id = books.publisher_id
            WHERE books.num_pages > 50
            GROUP BY publishers.publisher
            ORDER BY cnt_books desc
            LIMIT 1
'''
pd.io.sql.read_sql(query, con = engine)
```

**Вывод**: больше всего книг (42 шт) выпустило издательство Penguin Books - это довольно большое и старое издательство 

### 4. Автор с самой высокой средней оценкой книг (только книги с 50 и более оценками)

```python
query = '''
            SELECT authors.author,
            round(avg(ratings.rating),3) as avg_rat
            FROM books
            left join authors on authors.author_id = books.author_id
            left join ratings on ratings.book_id = books.book_id
            WHERE books.book_id IN (
                    SELECT book_id
                    FROM ratings
                    GROUP BY book_id
                    HAVING count(rating_id) >= 50
                    ORDER BY count(rating_id) desc)
            GROUP BY authors.author
            ORDER BY avg_rat desc
            LIMIT 1
'''
pd.io.sql.read_sql(query, con = engine)
```

**Вывод:** самая высокая средняя оценка книг у автора Дж.К. Роулинг. Средняя оценка ее книг составляет 4.287

### 5. Среднее количество обзоров от пользователей, которые поставили больше 50 оценок

```python
query = '''
            SELECT round(avg(review_cnt)) as cnt_rev
            FROM (
                SELECT username,
                count(review_id) as review_cnt
                FROM reviews
                WHERE username IN (
                                    SELECT username
                                    FROM ratings
                                    GROUP BY username
                                    HAVING count(rating_id) > 50)
                GROUP BY username
                ) as rev_count
'''
pd.io.sql.read_sql(query, con = engine)
```

**Вывод:** пользователи, поставившие более 50 оценок, в среднем пишут по 24 обзора
